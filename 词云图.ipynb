# 数据读入

```python
#读入文件2024.txt
text=open('2024.txt', 'r',encoding='utf-8').read().replace('\n','')

#查看已读入数据
text[:]
```

# 分词处理

```python
# 初始化 NLPIR 系统
pynlpir.open()
```

```python
# 调用 pynlpir.segment() 函数进行中文分词和词性标注  
# text: 要进行分词和词性标注的文本字符串  
# pos_names='parent': 词性标注使用父类的名称（即更宽泛的词性分类）  
# pos_english=False: 词性标注的输出不使用英文（即使用中文词性标注）
pynlpir.segment(text,pos_names='parent',pos_english=False)[:]
```

# 自定义词汇

```python
#使用 pynlpir 的 nlpir 模块添加自定义词汇到 NLPIR 的用户词典中
pynlpir.nlpir.AddUserWord('幸福'.encode('utf8'),'adjective')
pynlpir.nlpir.AddUserWord('前所未有'.encode('utf8'),'adjective')
pynlpir.nlpir.AddUserWord('整齐划一'.encode('utf8'),'adjective')
pynlpir.nlpir.AddUserWord('脱贫攻坚'.encode('utf8'),'noun')
pynlpir.nlpir.AddUserWord('独龙江'.encode('utf8'),'noun')
```

```python
#重新分词
pynlpir.segment(text,pos_names='parent',pos_english=False)[:]
```

```python
#将分词和词性标注的结果存储在 words_with_pos 变量中
words_with_pos = pynlpir.segment(text, pos_names='parent', pos_english=False)
```

```python
#将分词和词性写入文件分词结果.txt
with open(r"C:\Users\13621\Desktop\分词结果.txt", 'w', encoding='utf-8') as f:  
    for word, pos in words_with_pos:  
        f.write(f"{word}/{pos} \n")  
    f.write("\n")
```

```python
#对文章数据进行分词，并追加至一个新的列表words
words = []
year=2024
year_words = []
year_words.extend(pynlpir.segment(text,pos_names='parent',pos_english=False))     
for j in range(len(year_words)):
    ls_year_words=list(year_words[j])
    ls_year_words.append(year)
    words.append(ls_year_words)
```

```python
#创建一个DataFrame，其中每一行都是一个 (词汇, 词性, 年份) 的元组
df_words = pd.DataFrame(words,columns=["词汇","词性","年份"])

#显示DataFrame的前25行  
df_words.head(31)
```

# 停用词处理

```python
#打开并读取名为 'stopwords.txt' 的文件
stopwords= open('stopwords.txt').read()
stopwords[:]
```

```python
#过滤停用词
for i in range(df_words.shape[0]):
    if(df_words.词汇[i] in stopwords):
        df_words.drop(i,inplace=True)
  
    else:
        pass
```

# 词性分布分析

```python
#统计2024文章的词性分布
df_WordSpeechDistribution = pd.DataFrame(df_words['词性'].value_counts(ascending=False))
df_WordSpeechDistribution.head(10)
```

```python
#修改列名
#.rename（）函数的参数：
    #inplace:是否修改数据框本身
    #columns={"旧名称1": "新名称1", "旧名称2": "新名称2"}
df_WordSpeechDistribution.rename(columns={'词性':'频数'},inplace=True)

#查看 df_WordSpeechDistribution DataFrame 的前 20 行
df_WordSpeechDistribution.head(20)
```

```python
# 增设一个新列“百分比”，并只显示前10行
df_WordSpeechDistribution['比例'] = df_WordSpeechDistribution['频数'] / df_WordSpeechDistribution['频数'].sum()
df_WordSpeechDistribution.head(10)
```

```python
#绘制前10位词性类别分布图
plt.rcParams["font.family"] = 'simHei'
plt.subplots(figsize=(7,5))
plt.rcParams["font.family"] ='STFangsong'
df_WordSpeechDistribution.iloc[:10]['频数'].plot(kind='barh')
plt.yticks(fontproperties=font,size=10)
plt.xlabel('频数',fontproperties=font,size=10)
plt.ylabel('词性',fontproperties=font,size=10)
plt.title('中华民族一家亲 同心共筑中国梦【述评】词性分布分析',fontproperties=font)
```

# 高频词分析

```python
# 定义6个主要词性列表
columns_slected=['动词','动词计数','名词','名词计数','代词','代词计数','量词','量词计数','副词','副词计数','形容词','形容词计数']

# 定义6类词性统计数据框
df_Top6 = pd.DataFrame(columns=columns_slected)

# 统计6大词性类别的具体内容，并保存到数据框
for i in range(0,12,2):
    df_Top6[columns_slected[i]] = df_words.loc[df_words['词性']==columns_slected[i]]['词汇'].value_counts().reset_index()['index']
    df_Top6[columns_slected[i+1]] = df_words.loc[df_words['词性']==columns_slected[i]]['词汇'].value_counts().reset_index()['词汇']
df_Top6.head(16)
```

# 词频统计

```python
#显示 df_words DataFrame 的前 15 行
df_words.head(15)
```

```python
#将df_words DataFrame中的"年份"列作为列标签，并将"词汇"列的值作为这些列下的数据
df_AnnualWords=df_words[["年份","词汇"]].pivot(columns="年份", values="词汇")
df_AnnualWords.head()
```

```python
#将df_AnnualWords DataFrame中的所有NaN值替换为0
df_AnnualWords.fillna(0,inplace=True)  
df_AnnualWords.head()
```

```python
#建一个新的DataFrame df_AnnualTopWords，来存储2024年出现频率最高的词汇
df_AnnualTopWords=pd.DataFrame(columns=[2024])
df_AnnualTopWords[2024]=df_AnnualWords[2024].value_counts().reset_index()["index"]
df_AnnualTopWords.index=df_AnnualTopWords.index+1
df_AnnualTopWords.head(20)
```

# 关键词分析

```python
#从df_AnnualTopWords[2024]列中的词汇中提取关键词，并将这些关键词存储在一个新的DataFrame df_annual_keywords中
df_annual_keywords = pd.DataFrame(columns=[2024])
df_annual_keywords[2024]=pynlpir.get_key_words(' '.join(df_AnnualTopWords[2024].astype('str')))
df_annual_keywords.head(10)
```
