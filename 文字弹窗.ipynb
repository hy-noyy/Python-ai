```python
import tkinter as tk
import random

class PopupManager:
    def __init__(self):
        self.root = tk.Tk()
        self.root.withdraw()  # 隐藏主窗口
        self.popups = []  # 存储所有弹窗引用
        self.sentences = [
            # 原始句子
            '我曾经喜欢过你，讨厌过你，埋怨过你，支持过你，和你一起走过了三千多章的故事',
            '天上的星星都掉下来了，你又在哪里看着我呢?',
            '众赫啊，我们能拯救世界你知道吗?',
            '独子有独子的人生。',
            '谨献给身在世界尽头，再也没有故事可读的你。',
            '我是刘众赫。',
            '你采取行动，我守望着那样的你。',
            '我就在这里，如此，你还是只盯着我阅读。',
            '再读一遍吧，独子。',
            '传说[生死与共的伙伴]继续讲述故事。',
            '您的■■为[永恒]。',
            '我们都是彼此的异界神格，',
            '在你的小说里，我感觉不到你。',
            '化身刘众赫，拒绝回归。',
            '愿故事庇佑你。',
            '这是只为了唯一一位读者所写的故事。',
            '只要我继续回归，总有一天会在遇到你吗?',
            
            # 新增句子
            '我一看到他，就想把他庇护在我的羽翼之下。',
            '化身金独子将死于挚爱之手。',
            '我们还会再见面吗?',
            '星座[隐秘的谋略家]的■■是[最古老的梦]。',
            '这是一个很棒的故事不是吗?。',
            '你妈妈好像不喜欢我。',
            '金独子对这个世界来说是必要的，我需要他。',
            '他将是我的伙伴，他将看到这一切的结束。',
            '你的命是我的。',
            '金独子，你知道吗，你的故事曾几何时拯救过我。',
            '“我来帮你结束你的故事。”',
            '谢谢你在我身边空无一人时陪伴我。',
            '金独子是我的生命。',
            '谢谢你改变了我的生命，众赫啊。',
            '这是你应得的爱。',
            
            # 最新添加的句子
            '现在，我们要么一起活，要么一起死。',
            '刘众赫无可避免地承认，金独子是他的一部分。'
        ]
        
        # 准备句子列表（每个句子出现三次）
        self.all_sentences = []
        for sentence in self.sentences:
            self.all_sentences.append(sentence)
            self.all_sentences.append(sentence)
            self.all_sentences.append(sentence)  # 每个句子添加三次
        
        # 随机打乱句子顺序，确保每三个弹窗中没有重复
        self.shuffle_with_no_triple_repeats()
        
        # 计数器
        self.current_index = 0
        self.total_popups = len(self.all_sentences)
    
    def shuffle_with_no_triple_repeats(self):
        """随机打乱句子顺序，确保每三个弹窗中没有重复"""
        # 首先随机打乱整个列表
        random.shuffle(self.all_sentences)
        
        # 检查并修复连续三个相同句子的情况
        n = len(self.all_sentences)
        for i in range(n - 2):
            if (self.all_sentences[i] == self.all_sentences[i+1] == self.all_sentences[i+2]):
                # 找到下一个不同的句子
                for j in range(i+3, n):
                    if self.all_sentences[j] != self.all_sentences[i]:
                        # 交换位置
                        self.all_sentences[i+2], self.all_sentences[j] = self.all_sentences[j], self.all_sentences[i+2]
                        break
        
        # 检查所有可能的三连组合
        for i in range(n - 2):
            # 检查是否有重复（不一定是三个相同，而是任意两个相同）
            if (self.all_sentences[i] == self.all_sentences[i+1] or 
                self.all_sentences[i] == self.all_sentences[i+2] or 
                self.all_sentences[i+1] == self.all_sentences[i+2]):
                
                # 找到下一个完全不同的句子
                for j in range(i+3, n):
                    if (self.all_sentences[j] != self.all_sentences[i] and 
                        self.all_sentences[j] != self.all_sentences[i+1] and 
                        self.all_sentences[j] != self.all_sentences[i+2]):
                        # 交换位置
                        self.all_sentences[i+2], self.all_sentences[j] = self.all_sentences[j], self.all_sentences[i+2]
                        break
    
    def create_first_popup(self):
        """创建并显示第一个弹窗（金独子），标题为'第四面墙'"""
        popup = tk.Toplevel(self.root)
        popup.title("第四面墙")
        popup.attributes("-topmost", True)
        
        # 居中显示
        screen_width = popup.winfo_screenwidth()
        screen_height = popup.winfo_screenheight()
        x = (screen_width - 250) // 2
        y = (screen_height - 120) // 2
        popup.geometry(f"250x120+{x}+{y}")
        
        # 创建框架用于居中内容
        frame = tk.Frame(popup)
        frame.place(relx=0.5, rely=0.5, anchor="center")
        
        # 添加文本标签（居中显示）
        label = tk.Label(frame, text="金独子", font=("微软雅黑", 14))
        label.pack()
        
        # 添加按钮（文字改为"阅读"）
        button = tk.Button(frame, text="阅读", command=lambda: self.close_first_popup(popup))
        button.pack(pady=5)
        
        # 绑定窗口关闭事件
        popup.protocol("WM_DELETE_WINDOW", lambda: self.close_first_popup(popup))
    
    def close_first_popup(self, popup):
        """关闭第一个弹窗并开始创建后续弹窗"""
        popup.destroy()
        self.create_remaining_popups()
    
    def create_popup(self, message):
        """创建一个弹窗，标题为'白日幽会'，文字居中显示"""
        popup = tk.Toplevel(self.root)
        popup.title("白日幽会")
        popup.attributes("-topmost", True)
        
        # 设置随机位置
        screen_width = popup.winfo_screenwidth()
        screen_height = popup.winfo_screenheight()
        x = random.randint(0, max(0, screen_width - 300))
        y = random.randint(0, max(0, screen_height - 150))
        popup.geometry(f"300x150+{x}+{y}")
        
        # 创建框架用于居中内容
        frame = tk.Frame(popup)
        frame.place(relx=0.5, rely=0.5, anchor="center")
        
        # 添加文本标签（居中显示）
        label = tk.Label(frame, text=message, wraplength=280,
                        font=("微软雅黑", 9), justify="center")
        label.pack(padx=10, pady=10)
        
        # 存储弹窗引用
        self.popups.append(popup)
        return popup
    
    def create_remaining_popups(self):
        """创建剩余的弹窗（每0.1秒一个）"""
        if self.current_index < self.total_popups:
            # 创建当前弹窗
            message = self.all_sentences[self.current_index]
            self.create_popup(message)
            self.current_index += 1
            
            # 0.1秒后创建下一个弹窗
            self.root.after(100, self.create_remaining_popups)
        else:
            # 所有弹窗创建完毕，3秒后关闭所有弹窗
            self.root.after(3000, self.close_all_popups)
    
    def close_all_popups(self):
        """关闭所有弹窗并退出程序"""
        for popup in self.popups:
            try:
                popup.destroy()
            except:
                pass
        self.root.destroy()
    
    def run(self):
        """运行程序"""
        self.create_first_popup()
        self.root.mainloop()

# 运行程序
if __name__ == "__main__":
    manager = PopupManager()
    manager.run()
```
